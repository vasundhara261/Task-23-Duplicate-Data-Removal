```python
import pandas as pd

# Create a sample dataset with duplicate records
data = {
    'Name': ['Anu', 'Rahul', 'Priya', 'Anu', 'Rahul'],
    'Age': [20, 21, 22, 20, 21],
    'City': ['Bangalore', 'Mumbai', 'Delhi', 'Bangalore', 'Mumbai']
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
print(df)

# Count rows before cleaning
before_rows = df.shape[0]

# Identify duplicate records
duplicates = df.duplicated()

print("\nDuplicate Records:")
print(df[duplicates])

# Count duplicate records
duplicate_count = duplicates.sum()
print("\nNumber of duplicate records:", duplicate_count)

# Remove duplicates
cleaned_df = df.drop_duplicates()

# Count rows after cleaning
after_rows = cleaned_df.shape[0]

print("\nCleaned Dataset:")
print(cleaned_df)

# Compare row counts
print("\nRows before cleaning:", before_rows)
print("Rows after cleaning:", after_rows)
```
